```python
import random
import matplotlib.pyplot as plt
```

```python
# ==========================
#  CLASSE AGENTE (Pessoa)
# ==========================
class Person:
  def __init__(self, id):
    self.id = id
    self.opiniao = random.uniform(0, 1)   # 0 = extremo A, 1 = extremo B
    self.influencia = random.uniform(0.1, 1.0)  # quanto afeta os outros
    self.abertura = random.uniform(0.1, 1.0)    # quanto aceita mudar
    self.hist_opiniao = [self.opiniao]

  def interagir(self, outro):
    """
    Um agente influencia o outro.
    """
    # diferença de opinião entre os dois
    diff = outro.opiniao - self.opiniao

    # influência mútua (autoajuste)
    self.opiniao += diff * self.abertura * outro.influencia * 0.1
    outro.opiniao -= diff * outro.abertura * self.influencia * 0.1

    # mantém dentro do intervalo 0–1
    self.opiniao = min(max(self.opiniao, 0), 1)
    outro.opiniao = min(max(outro.opiniao, 0), 1)

    # registra histórico
    self.hist_opiniao.append(self.opiniao)
    outro.hist_opiniao.append(outro.opiniao)
```

```python
# ==========================
#  CLASSE SOCIEDADE
# ==========================
class Society:
  "Gera as pessoas usando a classe Person"
  def __init__(self, n_pessoas=50):
    self.pessoas = [Person(i) for i in range(n_pessoas)]
    self.step = 0
    self.hist_media = []

  def rodada(self):
    """
    Uma rodada de interações entre agentes.
    """
    for _ in range(len(self.pessoas)):
        a, b = random.sample(self.pessoas, 2)
        a.interagir(b)

    # média das opiniões no tempo
    media = sum(p.opiniao for p in self.pessoas) / len(self.pessoas)
    self.hist_media.append(media)
    self.step += 1

  def simular(self, n_passos=100):
    for _ in range(n_passos):
        self.rodada()

  def plotar(self):
    """
    Mostra gráficos da evolução.
    """
    plt.figure(figsize=(10,5))
    plt.plot(self.hist_media, label="Média da sociedade")
    plt.title("Evolução da opinião média")
    plt.xlabel("Tempo (iterações)")
    plt.ylabel("Opinião média (0 a 1)")
    plt.legend()
    plt.grid(True)
    plt.show()


```

```python
# ==========================
#  EXECUÇÃO
# ==========================
if __name__ == "__main__":
  sociedade = Society(n_pessoas=80)
  sociedade.simular(n_passos=200)
  sociedade.plotar()

  # Exemplo: ver opiniões finais
  print("\nOpiniões finais:")
  for p in sociedade.pessoas[:10]:
      print(f"Agente {p.id}: {p.opiniao:.2f}")
```
